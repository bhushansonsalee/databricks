```python
from pyspark.sql.functions import (
    col, trim, ltrim, rtrim, lower, upper, initcap,
    length, lpad, rpad, substring, concat, lit
)

dirty = spark.createDataFrame([
    (1, "  aarav MEHTA  ", "  Aarav.Mehta@Gmail.COM  "),
    (2, "PRIYA sharma",   "priya@COMPANY.co.in"),
    (3, " rohan  PATEL ", "ROHAN.patel@yahoo.com "),
], ["id", "name", "email"])

clean = dirty.select(
    col("id"),

    # ── Trim + Proper Case ──
    initcap(trim(col("name"))).alias("name"),

    # ── Trim + Lowercase email ──
    lower(trim(col("email"))).alias("email"),

    # ── Extract username from email ──
    substring(
        lower(trim(col("email"))),
        1,
        # find position of '@' minus 1
        locate("@", lower(trim(col("email")))) - 1
    ).alias("username"),

    # ── Pad employee ID to 6 digits ──
    lpad(col("id").cast("string"), 6, "0").alias("emp_code"),
)

clean.show(truncate=False)
```
