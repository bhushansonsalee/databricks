```sql
%sql
CREATE TABLE employee_n (
  emp_id INT,
  emp_name VARCHAR(50),
  dept VARCHAR(20),
  salary INT,
  hire_date DATE
);
```

```sql
%sql
SELECT emp_name, dept, salary,
ROW_NUMBER() OVER (PARTITION BY dept ORDER BY salary DESC) rn 
FROM employee_n;
```

```sql
%sql
WITH cte AS (
    SELECT *,
           ROW_NUMBER() OVER (PARTITION BY dept ORDER BY salary DESC) rn
    FROM employee_n
)
SELECT *
FROM cte
WHERE rn <= 2;
```

```sql
SELECT emp_name, dept,
SUM(salary) OVER (PARTITION BY dept) AS dept_total
FROM employee_n;
```
