```python
data = [
(1,"John","Pune",2024),
(2,"Sara","Mumbai",2024),
(3,"Mike","Delhi",2024),
(4,"Alice","Pune",2025),
(5,"Bob","Mumbai",2025),
(6,"Tom","Delhi",2025)
]

columns = ["id","name","city","year"]

df = spark.createDataFrame(data,columns)

df.show()
```

```python
hash_df = df.repartition(3,"id")

hash_df.explain(True)
```

```python
range_df = df.repartitionByRange(3,"id")

range_df.explain(True)
```

```python
rr_df = df.repartition(3)

rr_df.explain(True)
```
