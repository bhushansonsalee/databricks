```python
from pyspark.sql.functions import rand

df = spark.range(0, 1000000).withColumn("user_id", (rand()*10000).cast("int")) \
                            .withColumn("region", (rand()*100).cast("int")) \
                            .withColumn("event_type", (rand()*50).cast("int"))

df.write.format("delta").mode("overwrite").saveAsTable("p_01_bsg.default.events")
```

```python
%sql
CREATE TABLE zorder_demo
USING DELTA
LOCATION 'p_01_bsg.default.events';
```
